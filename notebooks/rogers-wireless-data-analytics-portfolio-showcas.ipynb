### SQL Example — CRTC Unit Standardization

The following SQLite query was used to standardize CRTC reporting units while preserving the original quarterly KPI table.

```sql
CREATE VIEW CRTC_Quarterly_KPIs_Standardized AS
SELECT
    Quarter,
    Total_Revenue / 1000000.0 AS Total_Revenue_CAD_Millions,
    Equipment_Revenue / 1000000.0 AS Equipment_Revenue_CAD_Millions,
    Service_Revenue / 1000000.0 AS Service_Revenue_CAD_Millions,
    Subscribers / 1000.0 AS Subscribers_Thousands,
    ARPU_With_Devices,
    ARPU_Without_Devices
FROM CRTC_Quarterly_KPIs;
```

**Result:** Standardized revenue to CAD millions and subscriber counts to thousands across 21 CRTC quarters. Reverse reconciliation confirmed 0 conversion mismatches.

*This SQL transformation was executed locally in DB Browser for SQLite. The Kaggle notebook uses the resulting exported CSV.*

### Analysis 1 — Wireless Service Revenue Growth

**Business Question:** Did Rogers' wireless service revenue growth follow the same trend as the broader Canadian mobile industry?

**Method:** Compare quarterly year-over-year (YoY) service revenue growth rates for Rogers and CRTC.

**Why YoY Growth?** Comparing each quarter with the same quarter in the previous year helps reduce seasonal effects and highlights changes in revenue growth momentum.

**Visualization:** A two-line chart showing Rogers and CRTC quarterly revenue growth percentages.

**Interpretation:** Differences between the two series will be examined as directional performance gaps, not definitive evidence of market-share changes or their causes.

````python

# Inspect available revenue growth columns
revenue_columns = [
    col for col in master.columns
    if "Revenue" in col or "Quarter" in col
]

print("AVAILABLE REVENUE COLUMNS")
for col in revenue_columns:
    print(col)
````

````python

import matplotlib.pyplot as plt
import pandas as pd

# 1. Select the relevant columns
revenue_trend = master[
    [
        "Quarter",
        "Rogers_Revenue_YoY_Pct",
        "CRTC_Revenue_YoY_Pct"
    ]
].copy()

# 2. Convert values to numeric and remove unavailable comparisons
for col in ["Rogers_Revenue_YoY_Pct", "CRTC_Revenue_YoY_Pct"]:
    revenue_trend[col] = pd.to_numeric(
        revenue_trend[col], errors="coerce"
    )

revenue_trend = revenue_trend.dropna(
    subset=["Rogers_Revenue_YoY_Pct", "CRTC_Revenue_YoY_Pct"]
)

# 3. Plot the two growth-rate series
plt.figure(figsize=(13, 5))

plt.plot(
    revenue_trend["Quarter"],
    revenue_trend["Rogers_Revenue_YoY_Pct"],
    marker="o",
    linewidth=2,
    label="Rogers",
    color="#E87722"
)

plt.plot(
    revenue_trend["Quarter"],
    revenue_trend["CRTC_Revenue_YoY_Pct"],
    marker="o",
    linewidth=2,
    label="Canadian Wireless Industry (CRTC)",
    color="#2675B8"
)

plt.axhline(0, color="gray", linewidth=1, linestyle="--")

plt.title("Wireless Revenue Growth: Rogers vs. CRTC")
plt.xlabel("Quarter")
plt.ylabel("Year-over-Year Growth (%)")
plt.xticks(rotation=45, ha="right")
plt.legend()
plt.grid(alpha=0.2)
plt.tight_layout()

plt.show()

# 4. Preview the values behind the chart
display(revenue_trend.tail(8).round(2))
````

The chart reveals a substantial slowdown in wireless revenue growth for both Rogers and the broader Canadian mobile industry.

**Key Observations**

- **2022–2023:** Rogers experienced stronger fluctuations than the industry, including a pronounced growth peak in 2023 Q3. The 2023 Shaw acquisition is an important comparability consideration.
- **2024:** Growth moderated for both series. In 2024 Q2, Rogers recorded 3.54% YoY growth versus 2.26% for CRTC; by 2024 Q3, growth had slowed to 1.97% and 0.34%, respectively.
- **2025–2026 Q1:** Both series approached zero growth, with some quarters recording negative YoY changes.

**Business Interpretation**

The slowdown was not unique to Rogers. However, differences in growth rates suggest that company-specific performance and industry-wide conditions should be examined separately.

**Limitation**

Rogers and CRTC revenue measures have different reporting scopes. The chart provides a directional benchmark rather than a direct market-share comparison. The 2023 acquisition also complicates historical comparisons.

### Analysis 2 — Subscriber Growth: Volume vs. Revenue

**Business Question:** Did Rogers continue to expand its wireless subscriber base while revenue growth slowed?

**Method:** Compare Rogers' quarterly subscriber growth with Canadian industry trends using year-over-year (YoY) rates.

**Why Subscriber Growth?**

Subscriber growth measures the expansion of the customer base, while revenue growth reflects overall monetization. Examining both indicators helps determine whether revenue performance is driven primarily by customer volume or whether revenue per subscriber requires further investigation.

**Tools:** Python (Pandas) for selecting and checking KPI data; Matplotlib for comparing subscriber growth trends.

**Expected Output:** A two-line chart comparing Rogers and CRTC subscriber growth across available quarters.

**Comparability Note:** Rogers and CRTC use different subscriber classifications. This analysis compares reported growth patterns directionally, rather than assuming identical customer populations.

````python

import pandas as pd
import matplotlib.pyplot as plt

# 1. Inspect subscriber-related columns
subscriber_columns = [
    col for col in master.columns
    if "Subscriber" in col
]

print("AVAILABLE SUBSCRIBER COLUMNS")
print("\n".join(subscriber_columns))

# 2. Select quarterly subscriber growth indicators
subscriber_trend = master[
    [
        "Quarter",
        "Rogers_Subscribers_YoY_Pct",
        "CRTC_Subscribers_YoY_Pct"
    ]
].copy()

# 3. Ensure growth rates are numeric
for col in [
    "Rogers_Subscribers_YoY_Pct",
    "CRTC_Subscribers_YoY_Pct"
]:
    subscriber_trend[col] = pd.to_numeric(
        subscriber_trend[col],
        errors="coerce"
    )

# 4. Retain comparable quarters
subscriber_trend = subscriber_trend.dropna(
    subset=[
        "Rogers_Subscribers_YoY_Pct",
        "CRTC_Subscribers_YoY_Pct"
    ]
)

# 5. Plot subscriber growth
plt.figure(figsize=(13, 5))

plt.plot(
    subscriber_trend["Quarter"],
    subscriber_trend["Rogers_Subscribers_YoY_Pct"],
    marker="o",
    linewidth=2,
    color="#E87722",
    label="Rogers"
)

plt.plot(
    subscriber_trend["Quarter"],
    subscriber_trend["CRTC_Subscribers_YoY_Pct"],
    marker="o",
    linewidth=2,
    color="#2675B8",
    label="Canadian Wireless Industry (CRTC)"
)

plt.axhline(0, color="gray", linestyle="--", linewidth=1)

plt.title("Subscriber Growth: Rogers vs. CRTC")
plt.xlabel("Quarter")
plt.ylabel("Year-over-Year Growth (%)")
plt.xticks(rotation=45, ha="right")
plt.legend()
plt.grid(alpha=0.2)
plt.tight_layout()
plt.show()

# 6. Display recent quarterly values
display(subscriber_trend.tail(8).round(2))
````

The subscriber growth analysis shows that both Rogers and the Canadian wireless industry maintained positive year-over-year subscriber growth throughout the comparable period, even as revenue growth slowed.

**Key Observations**

- **2022–2023:** Rogers experienced relatively strong subscriber growth, with a pronounced increase around the 2023 Shaw acquisition period.
- **2024:** Subscriber growth moderated. In 2024 Q3, Rogers recorded 2.15% YoY growth versus 2.56% for the industry.
- **2025:** Rogers continued expanding its subscriber base. Subscriber growth reached 3.51% in Q1 and 3.46% in Q2, while reported revenue growth was only 1.50% and 0.55%, respectively.

**Business Interpretation**

Subscriber growth remained positive, but revenue did not increase proportionally. This suggests that subscriber expansion alone cannot fully explain Rogers' revenue performance.

The difference raises an important question about customer monetization: whether average revenue per user declined enough to offset part of the benefit from subscriber growth.

**Comparability Considerations**

The 2023 Shaw acquisition affects historical comparisons, and Rogers and CRTC subscriber definitions are not identical. These results describe reported growth trends rather than proving organic customer growth or market-share changes.

Did declining ARPU offset the positive contribution from subscriber growth?

### Analysis 3 — ARPU: Understanding Customer Monetization

**Business Question:** Did declining average revenue per user (ARPU) contribute to Rogers' slowing wireless revenue growth?

**Method:** Examine quarterly ARPU year-over-year growth rates for Rogers and the Canadian wireless industry.

**Why ARPU?**

ARPU measures average monthly revenue per subscriber. An expanding subscriber base does not necessarily lead to proportional revenue growth if average revenue per subscriber declines.

**Tools:** Python (Pandas) for selecting and validating the data; Matplotlib for visualizing ARPU trends.

**Expected Output:** A two-line chart showing Rogers and CRTC ARPU YoY growth, followed by a table of recent quarterly observations.

**Comparability Note:** ARPU definitions, revenue components and subscriber populations may differ between Rogers and CRTC. The comparison is directional and does not establish customer-level causes.

````python

import pandas as pd
import matplotlib.pyplot as plt

# 1. Identify ARPU-related columns
arpu_columns = [
    col for col in master.columns
    if "ARPU" in col
]

print("AVAILABLE ARPU COLUMNS")
for col in arpu_columns:
    print(col)

# 2. Select ARPU year-over-year growth data
arpu_trend = master[
    [
        "Quarter",
        "Rogers_ARPU_YoY_Pct",
        "CRTC_ARPU_YoY_Pct"
    ]
].copy()

# 3. Convert values to numeric
for col in ["Rogers_ARPU_YoY_Pct", "CRTC_ARPU_YoY_Pct"]:
    arpu_trend[col] = pd.to_numeric(
        arpu_trend[col],
        errors="coerce"
    )

# 4. Keep quarters with both ARPU growth rates
arpu_trend = arpu_trend.dropna(
    subset=["Rogers_ARPU_YoY_Pct", "CRTC_ARPU_YoY_Pct"]
)

# 5. Plot ARPU growth trends
plt.figure(figsize=(13, 5))

plt.plot(
    arpu_trend["Quarter"],
    arpu_trend["Rogers_ARPU_YoY_Pct"],
    marker="o",
    linewidth=2,
    color="#E87722",
    label="Rogers"
)

plt.plot(
    arpu_trend["Quarter"],
    arpu_trend["CRTC_ARPU_YoY_Pct"],
    marker="o",
    linewidth=2,
    color="#2675B8",
    label="Canadian Wireless Industry (CRTC)"
)

plt.axhline(
    0,
    color="gray",
    linestyle="--",
    linewidth=1
)

plt.title("ARPU Growth: Rogers vs. CRTC")
plt.xlabel("Quarter")
plt.ylabel("Year-over-Year ARPU Growth (%)")
plt.xticks(rotation=45, ha="right")
plt.legend()
plt.grid(alpha=0.2)
plt.tight_layout()

plt.show()

# 6. Display recent results
display(arpu_trend.tail(8).round(2))
````

### Analysis 4 — Revenue Bridge: Quantifying Revenue Drivers

**Business Question:** How much did subscriber growth and ARPU changes contribute to Rogers' year-over-year wireless service revenue performance in 2025?

**Analytical Method: Revenue Driver Decomposition**

The revenue bridge separates service revenue movements into three components:

| Component | Meaning |
|---|---|
| **Subscriber Effect** | Estimated revenue contribution from changes in subscriber volume |
| **ARPU Effect** | Estimated revenue contribution from changes in average revenue per subscriber |
| **Residual** | Reconciliation adjustment between modeled and reported revenue changes |

**Why This Method?**

Trend charts show that subscriber growth and ARPU moved in opposite directions, but they do not quantify their respective contributions. A revenue bridge helps translate those trends into measurable financial impacts.

**Tools & Responsibilities**

- **Python (Pandas):** Organize quarterly driver calculations and verify the revenue bridge.
- **Python (Matplotlib):** Visualize subscriber effects, ARPU effects and actual service revenue changes.
- **Tableau:** Present the completed bridge as an executive-facing dashboard.

**Validation Principle**

Subscriber Effect + ARPU Effect + Residual = Actual YoY Service Revenue Change

**Scope:** Focus on the four quarters of 2025. Monetary values are expressed in CAD millions.

#### Revenue Bridge Visualization & Validation

The previous analysis identified a divergence between subscriber growth and ARPU performance. This step quantifies the modeled financial impact of each driver across the four quarters of 2025.

**Visualization Method**

A grouped bar chart compares four components: Subscriber Effect, ARPU Effect, Residual Effect, and Actual Service Revenue Change.

**Validation Method**

Python checks whether the sum of the three modeled components reconciles with the reported service revenue change.

**Expected Output**

A quarterly revenue-driver chart and a reconciliation table, with all monetary values expressed in CAD millions.

**Interpretation Note**

The residual is a mathematical adjustment used to reconcile the model with reported revenue. It does not establish a verified operational cause.

````python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Select the 2025 revenue bridge data
bridge_cols = [
    "Subscriber_Effect_CAD_M",
    "ARPU_Effect_CAD_M",
    "Residual_Effect_CAD_M",
    "Service_Revenue_Change_YoY_CAD_M"
]

bridge = master.loc[
    master["Quarter"].isin(
        ["2025 Q1", "2025 Q2", "2025 Q3", "2025 Q4"]
    ),
    ["Quarter"] + bridge_cols
].copy()

bridge = bridge.set_index("Quarter").reindex(
    ["2025 Q1", "2025 Q2", "2025 Q3", "2025 Q4"]
)

# 2. Convert the financial fields to numeric
bridge[bridge_cols] = bridge[bridge_cols].apply(
    pd.to_numeric, errors="coerce"
)

# 3. Recalculate the bridge reconciliation
bridge["Calculated_Change"] = (
    bridge["Subscriber_Effect_CAD_M"]
    + bridge["ARPU_Effect_CAD_M"]
    + bridge["Residual_Effect_CAD_M"]
)

bridge["Reconciliation_Difference"] = (
    bridge["Calculated_Change"]
    - bridge["Service_Revenue_Change_YoY_CAD_M"]
)

# 4. Show the actual quarterly results
print("2025 REVENUE BRIDGE — CAD MILLIONS")
display(bridge.round(2))

valid = bridge["Reconciliation_Difference"].notna().all()
if valid:
    passed = np.allclose(
        bridge["Reconciliation_Difference"],
        0,
        atol=0.02
    )
    print("Reconciliation passed:", passed)
else:
    print("Some bridge values are missing; review required.")

# 5. Create a grouped bar chart
fig, ax = plt.subplots(figsize=(12, 6))

x = np.arange(len(bridge))
width = 0.20

series = [
    ("Subscriber_Effect_CAD_M", "Subscriber Effect", "#2E8B57"),
    ("ARPU_Effect_CAD_M", "ARPU Effect", "#D9534F"),
    ("Residual_Effect_CAD_M", "Residual", "#999999"),
    ("Service_Revenue_Change_YoY_CAD_M", "Actual Change", "#2675B8")
]

for i, (column, label, color) in enumerate(series):
    ax.bar(
        x + (i - 1.5) * width,
        bridge[column],
        width,
        label=label,
        color=color
    )

ax.axhline(0, color="black", linewidth=0.8)
ax.set_xticks(x)
ax.set_xticklabels(bridge.index)
ax.set_ylabel("CAD Millions")
ax.set_title("Rogers Wireless Service Revenue Bridge — 2025")
ax.legend(ncol=2)
ax.grid(axis="y", alpha=0.2)

plt.tight_layout()
plt.show()
````
